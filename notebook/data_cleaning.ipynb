- Since the goal is to predict one day in advance, the data has been shifted by one row.

```python
surface_df_copy["TH"] = surface_df_copy["TH"].shift(-1)
surface_df_copy.tail()
```

```python
print(surface_df_copy["TH"].dtype)
```

```python
surface_df_copy["TH"] = pd.to_numeric(surface_df_copy["TH"], errors="coerce")
```

```python
surface_df_copy.loc[surface_df_copy["TH"].isna(), "TH"] = 0
surface_df_copy["TH"] = surface_df_copy["TH"].astype(int)
print(surface_df_copy.tail())
print(surface_df_copy["TH"].dtype)
```
